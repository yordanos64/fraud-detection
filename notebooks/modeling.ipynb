```python
import sys
sys.path.append('../')
from src.modelling_pipeline import prepare_and_split_data
import xgboost as xgb
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score, average_precision_score

# 1. Load data via modular src pipelines
X_train_f, X_test_f, y_train_f, y_test_f, preprocessor = prepare_and_split_data("../data/processed/fraud_final.csv", "class")
X_train_c, X_test_c, y_train_c, y_test_c, _ = prepare_and_split_data("../data/processed/creditcard_final.csv", "Class", is_credit=True)

# 2. Train E-Commerce Models
lr_fraud = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42).fit(X_train_f, y_train_f)
xgb_fraud = xgb.XGBClassifier(n_estimators=150, max_depth=5, learning_rate=0.1, random_state=42, eval_metric='logloss').fit(X_train_f, y_train_f)

# 3. Print E-Commerce Metrics to the screen
print("--- E-COMMERCE ENSEMBLE RESULTS ---")
print("AUC-PR:", average_precision_score(y_test_f, xgb_fraud.predict_proba(X_test_f)[:, 1]))

# 4. Train Bank Credit Card Models
lr_credit = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42).fit(X_train_c, y_train_c)
xgb_credit = xgb.XGBClassifier(n_estimators=150, max_depth=5, learning_rate=0.1, random_state=42, eval_metric='logloss').fit(X_train_c, y_train_c)

print("\n--- BANK CREDIT CARD ENSEMBLE RESULTS ---")
print("AUC-PR:", average_precision_score(y_test_c, xgb_credit.predict_proba(X_test_c)[:, 1]))
```
