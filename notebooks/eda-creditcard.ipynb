```python
import pandas as pd

# 1. Load Raw Bank Transactions
credit_df = pd.read_csv("../data/raw/creditcard.csv")

# 2. Check Completeness
print("Missing Matrix Count:", credit_df.isnull().sum().sum())
print("Duplicate Records:", credit_df.duplicated().sum())

# 3. Quantify Skewness
counts = credit_df['Class'].value_counts()
pcts = credit_df['Class'].value_counts(normalize=True) * 100
print(f"\nBank Imbalance Profile:\nLegit (0): {counts[0]} ({pcts[0]:.4f}%)")
print(f"Fraud (1): {counts[1]} ({pcts[1]:.4f}%)")

# 4. Save Copy Directly to Processed Folder
credit_df.to_csv("../data/processed/creditcard_final.csv", index=False)
print("\nTask 1 part 3 complete: Bank data profile generated and saved!")
```
